## Final Model Selection

**Selected Model:** Gradient Boosting (Tuned)

**Why this model was selected:**
1. **Superior Performance:** It achieved the highest Test Score and F1 Score, indicating the best balance between precision and recall.
2. **Generalization:** The gap between the Training Score (0.96) and Validation Score (0.91) is small, suggesting that the model generalizes well and is not severely overfitting.
3. **Reliability:** The Cross-Validation score (0.90) confirms that the performance is stable across different data splits.
4. **Business Impact:** In our case, the higher ROC-AUC (0.95) means the model is exceptionally good at distinguishing between the positive and negative classes, which is the primary goal of the project.

While the model has higher complexity and lower interpretability than Logistic Regression, the significant gain in accuracy justifies the trade-off.
