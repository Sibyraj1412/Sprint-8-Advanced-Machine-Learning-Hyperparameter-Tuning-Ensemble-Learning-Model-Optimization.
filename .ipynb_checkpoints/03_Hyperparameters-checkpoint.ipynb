# Notebook 3 – Hyperparameters

## Parameters vs Hyperparameters
In machine learning, it is crucial to distinguish between these two terms:

### 1. Parameters
Parameters are the configuration variables that the model **learns automatically** from the training data. They are internal to the model.
- **Example:** The weights ($\beta$) and intercept in Linear Regression, or the split points in a Decision Tree.
- **How they are set:** Through optimization algorithms (e.g., Gradient Descent).

### 2. Hyperparameters
Hyperparameters are the configuration variables that you **set manually** before the training process begins. They govern how the learning process occurs.
- **Example:** The learning rate, the depth of a tree, or the number of neighbors in KNN.
- **How they are set:** Through experimentation, intuition, or automated tuning (e.g., Grid Search).

### Why Hyperparameters Matter
Hyperparameters directly control the balance between **Overfitting** and **Underfitting**. 
- A hyperparameter that is too "loose" (e.g., very high `max_depth` in a tree) leads to overfitting.
- A hyperparameter that is too "strict" (e.g., very low `max_depth`) leads to underfitting.

## Common Hyperparameters across Models

### 1. Decision Tree
- `max_depth`: The maximum depth of the tree. Controls the complexity. Higher depth $\to$ potentially more overfitting.
- `min_samples_split`: The minimum number of samples required to split an internal node.
- `min_samples_leaf`: The minimum number of samples required to be at a leaf node.
- `criterion`: The function to measure the quality of a split (`"gini"` or `"entropy"`).

### 2. Random Forest
- `n_estimators`: The number of trees in the forest. More trees generally improve stability but increase computation time.
- `max_depth`: Limits the depth of each individual tree.
- `max_features`: The number of features to consider when looking for the best split.
- `min_samples_split`: Minimum samples required to split a node.

### 3. K-Nearest Neighbors (KNN)
- `n_neighbors`: The number of neighbors to consider ($K$). Controls the local vs global nature of the prediction.
- `weights`: Weight function used in prediction (`"uniform"` or `"distance"`).
- `metric`: The distance metric used (`"euclidean"`, `"manhattan"`, etc.).

### 4. Support Vector Machines (SVM)
- `C`: Regularization parameter. Trade-off between smooth decision boundary and classifying training points correctly.
- `kernel`: Specifies the kernel type (`"linear"`, `"poly"`, `"rbf"`, `"sigmoid"`). Defines the shape of the boundary.
- `gamma`: Kernel coefficient for `"rbf"`. Defines how far the influence of a single training example reaches.

### 5. Logistic Regression
- `C`: Inverse of regularization strength. Smaller values $\to$ stronger regularization.
- `penalty`: Specifies the norm used in regularization (`"l1"`, `"l2"`, `"elasticnet"`, `None`).
- `solver`: Algorithm to use in the optimization problem (e.g., `"liblinear"`, `"lbfgs"`).
