# Notebook 11 – Advanced Cross Validation Strategies

## Introduction to Cross Validation (CV)
Cross validation is a technique used to evaluate how a model will generalize to an independent dataset. It prevents overfitting by ensuring the model is tested on different subsets of the data.

### 1. K-Fold Cross Validation
The data is split into $k$ equal folds. The model is trained $k$ times, each time using a different fold as the test set and the remaining $k-1$ folds as the training set.
- **When to use:** Standard datasets where samples are independent and identically distributed (i.i.d).

### 2. Stratified K-Fold
Similar to K-Fold, but it ensures that each fold has the same percentage of samples of each target class as the complete set.
- **When to use:** Imbalanced datasets (essential for classification).

### 3. Repeated K-Fold
K-Fold is repeated $n$ times with different random splits. This provides a more robust estimate of the model's performance.
- **When to use:** Small datasets where variance in a single K-Fold run might be high.

### 4. Leave-One-Out (LOOCV)
A special case of K-Fold where $k$ equals the number of samples. Every single sample is used as a test set once.
- **When to use:** Extremely small datasets. Very computationally expensive.

### 5. Group K-Fold
Ensures that the same group (e.g., all photos of the same patient) is never split between the training and testing sets.
- **When to use:** Data with groups/clusters (grouped data).

### 6. Time Series Split
Standard K-Fold fails for time series because it would use "future" data to predict the "past". Time Series Split uses a sliding or expanding window to respect chronological order.
- **When to use:** Stock prices, weather data, sensor logs.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import KFold, StratifiedKFold, RepeatedKFold, LeaveOneOut, TimeSeriesSplit, cross_val_score
from sklearn.ensemble import RandomForestClassifier

# Create dataset
X, y = make_classification(n_samples=200, n_features=10, weights=[0.8, 0.2], random_state=42)
model = RandomForestClassifier(random_state=42)

# 1. K-Fold
kf = KFold(n_splits=5, shuffle=True, random_state=42)
score_kf = cross_val_score(model, X, y, cv=kf)

# 2. Stratified K-Fold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
score_skf = cross_val_score(model, X, y, cv=skf)

# 3. Repeated K-Fold
rkf = RepeatedKFold(n_splits=5, n_repeats=3, random_state=42)
score_rkf = cross_val_score(model, X, y, cv=rkf)

# 4. Leave-One-Out
loo = LeaveOneOut()
# Note: LOOCV can be slow, using a subset for demo
score_loo = cross_val_score(model, X[:50], y[:50], cv=loo)

print(f"K-Fold Mean Accuracy: {np.mean(score_kf):.4f}")
print(f"Stratified K-Fold Mean Accuracy: {np.mean(score_skf):.4f}")
print(f"Repeated K-Fold Mean Accuracy: {np.mean(score_rkf):.4f}")
print(f"LOOCV Mean Accuracy (subset): {np.mean(score_loo):.4f}")
```

## Validation Strategy Selection Guide

| Dataset Type | Recommended Strategy | Reason |
| :--- | :--- | :--- |
| Standard, Balanced | **K-Fold** | Simple, efficient, reliable. |
| Imbalanced | **Stratified K-Fold** | Maintains class proportions. |
| Very Small | **LOOCV** | Maximizes training data per fold. |
| Grouped / Clustered | **Group K-Fold** | Prevents data leakage between groups. |
| Time-dependent | **Time Series Split** | Prevents look-ahead bias. |
