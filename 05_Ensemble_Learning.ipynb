# Notebook 5 – Ensemble Learning

## What is Ensemble Learning?
Ensemble Learning is a machine learning paradigm where multiple models (often called "base learners" or "weak learners") are trained to solve the same problem and their results are combined to produce a single, more accurate prediction.

### The Intuition: Why combine models?
The core idea is that **the collective intelligence of a group is better than that of a single individual**. 
In ML terms:
1. **Error Reduction:** Different models make different mistakes. By averaging or voting, the individual errors cancel each other out.
2. **Stability:** Ensemble models are typically less sensitive to noise in the training data compared to a single complex model.
3. **Generalization:** By combining models with different biases, the ensemble can capture a more robust pattern of the data.

## Types of Ensembles

### 1. Homogeneous Ensembles
All base learners are of the same type (e.g., all are Decision Trees).
- **Example:** Random Forest.

### 2. Heterogeneous Ensembles
Base learners are of different types (e.g., one KNN, one SVM, and one Decision Tree).
- **Example:** Voting Classifier.

### Common Ensemble Techniques:

- **Bagging (Bootstrap Aggregating):** Building multiple independent models in parallel and averaging their predictions. Focuses on reducing **Variance**.
- **Boosting:** Building models sequentially, where each new model attempts to correct the errors of the previous ones. Focuses on reducing **Bias**.
- **Voting:** A simple combination method.
    - *Hard Voting:* Majority class wins.
    - *Soft Voting:* Average of probabilities wins.
- **Stacking:** Using a "meta-model" to learn how to best combine the predictions of several base models.
- **Blending:** A variation of stacking that uses a hold-out validation set to train the meta-model to prevent leakage.
